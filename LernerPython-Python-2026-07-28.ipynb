# Enums

If you need/want a set of symbolic values, many languages offer the option of an "enum." The idea is that you could define a bunch of constants or variables, but it's easier to do it within a more established type.



```python
from enum import Enum

class Days(Enum):
    SUNDAY = 1
    MONDAY = 2
    TUESDAY = 3
    WEDNESDAY = 4
    THURSDAY = 5
    FRIDAY = 6
    SATURDAY = 7      
    
    
```

With this class, and with its class attributes, I can now talk about `Days.SUNDAY`, and have a value that's basically a constsant.

You could think of enums as a namespaces for related constants.

```python
Days.SUNDAY
```

```python
Days.SUNDAY.value
```

```python
from operator import attrgetter

list_of_days = [Days.SUNDAY, Days.WEDNESDAY, Days.MONDAY]

sorted(list_of_days, key=attrgetter('value'))
```

```python
from enum import IntEnum

class Days(IntEnum):
    SUNDAY = 1
    MONDAY = 2
    TUESDAY = 3
    WEDNESDAY = 4
    THURSDAY = 5
    FRIDAY = 6
    SATURDAY = 7      
    
```

```python
list_of_days = [Days.SUNDAY, Days.WEDNESDAY, Days.MONDAY]

sorted(list_of_days)
```

```python
# automatically number!

from enum import IntEnum, auto

class Days(IntEnum):
    SUNDAY = 0
    MONDAY = auto()
    TUESDAY = auto()
    WEDNESDAY = auto()
    THURSDAY = auto()
    FRIDAY = auto()
    SATURDAY = auto()   
    
```

```python
Days.SUNDAY
```

```python
Days.MONDAY
```

```python

class Months(IntEnum):
    JANUARY = 0
    FEBRUARY = auto()
    MARCH = auto()
```

```python
Months.JANUARY
```

```python
Months.FEBRUARY
```

# Pathlib

If you want to work with directories and files, you have a *lot* of functions you can use, most of which are in the `os` module (in the standard library), but some of which are scattered elsewhere. For example, I love to use `glob`.

`pathlib` is from Python 3.4, and it provides one class, `Path`. That class can be used for any file or directory on any operating system. It takes all of the functionality that was in `os` and elsewhere, and makes that all into methods on a `Path` object.

The bottom line is that it centralizes all of the working with paths, and makes it easier.

```python
from pathlib import Path
```

```python
p = Path('/Users/reuven')
```

```python
type(p)
```

```python
p
```

```python
p.listdir()
```

```python
for one_path in p.iterdir():
    print(type(one_path))
```

When I create a `Path` on a directory, any action I take that lists files in that directory returns new `Path` objects, which means that they also have all of those methods.

```python
for one_path in p.glob('*.txt'):
    print(one_path)
```

```python
# what if I want to get a file in the Path?
# for example, I have a .zshrc file in my home directory. 
# How can I get a Path to that?

p / '.zshrc'
```

```python
p / Path('/Users/reuven/.zshrc')
```

```python
p.is_dir()
```

```python
p.is_file()
```

# Exercise: File sizes

1. Ask the user to enter a globbing pattern.
2. Go through each file that matches that pattern, and only keep "real" files -- that is, something for which `is_file` returns `True`
4. Based on this, create a dict in which the keys are filenames and the values are the file sizes.
5. Print them, sorted by size (lowest to highest)

```python
from pathlib import Path

dirname = input('Enter dirname: ').strip()
pattern = input('Enter pattern: ').strip()

p = Path(dirname)        # create the Path object
files = p.glob(pattern)  # get files matching this pattern

all_files_dict = {}

for one_file in files:
    if one_file.is_file():
        size = one_file.stat().st_size
        all_files_dict[one_file.name] = size

all_files_dict
```

```python
# how can we sort this?

import operator

sorted(all_files_dict.items(), key=operator.itemgetter(1))  # itemgetter(1) is the same as [1] applied to each pair
```

# Tools

I remember when `pip` didn't come with Python, and you had to download and install it yourself.

At a certain point, Python came with the `ensurepip` module, which if you invoked it, would ensure that `pip` was installed, and that you could then download and install packages from PyPI.

Then `pip` became standardized. And a bunch of other packaging and other tools came out.

Some of the most interesting tools were about improving your code quality.

- `pep8`, which was then renamed `pycodestyle`, would evaluate your code as per PEP 8 instructions
- `pylint`, which checked your code quality
- `pyflakes`, which also checked your code quality

The good news is that these tools would give you decent feedback on your code.

The bad news:

- No one remembered which did what, so we ran all of them
- They were somewhat inconsistent
- They were slow



# `black` came out

The idea was that you would run `black` against your code, and it would reformat it!

- It would force your code to adhere to PEP8 standards

The idea was that if everyone used Black, then everyone's code would like closer to each others.

It was strongly opinionated -- one opinion was that "" are better than ''.

# Ruff came out

Ruff (from Astral) came out 3-4 years ago. It started as just a linter, something that would point to problems in your code.

Ruff would go through your code and tell you where you had issues:

- Unused variables
- Unused imports

The biggest thing about Ruff was that it was written in Rust, and thus ran blazingly fast.
